# PHÂN TÍCH MỨC ĐỘ CẠNH TRANH VÀ DỰ ĐOÁN ĐIỂM CHUẨN ĐẠI HỌC KHỐI NGÀNH CÔNG NGHỆ – KỸ THUẬT: NGHIÊN CỨU TẠI KHU VỰC HÀ NỘI (2023–2026)

## PHẦN 1. HIỂU BÀI TOÁN VÀ PHƯƠNG PHÁP TIẾP CẬN: DỰ ĐOÁN ĐIỂM CHUẨN & ĐÁNH GIÁ MỨC ĐỘ CẠNH TRANH

### 1.1. Bối cảnh thực tiễn và lý do thực hiện dự án

**Bối cảnh tuyển sinh khối ngành Công nghệ – Kỹ thuật**

Trong hệ thống tuyển sinh đại học tại Việt Nam, xét tuyển bằng điểm thi tốt nghiệp THPT vẫn là phương thức nền tảng quyết định cơ hội trúng tuyển của đông đảo thí sinh. Từ năm 2025, kỳ thi tốt nghiệp THPT chính thức áp dụng Chương trình Giáo dục phổ thông 2018 với cấu trúc 4 môn thi (Toán, Ngữ văn bắt buộc và 2 môn tự chọn), đồng thời lần đầu tiên đưa môn Tin học và Công nghệ vào kỳ thi, hình thành các tổ hợp xét tuyển mới cho khối kỹ thuật như `X06` (Toán – Lý – Tin) và `X26` (Toán – Anh – Tin) [1]. Trên quy mô 4.292.528 bản ghi điểm thi toàn quốc giai đoạn 2023–2026, mặt bằng điểm các môn chủ chốt (Toán, Vật lí, Tiếng Anh, Tin học) và số lượng thí sinh theo từng tổ hợp có sự dịch chuyển đáng kể qua các năm [2].

Tại khu vực Hà Nội — nơi quy tụ mạng lưới các trường đại học đào tạo kỹ thuật và công nghệ trọng điểm phía Bắc, hai mảng đào tạo cốt lõi gồm **Công nghệ thông tin (`IT`)** và **Kỹ thuật Điện tử – Vi mạch – Tự động hóa – Cơ khí (`ENG`)** luôn thu hút lượng lớn thí sinh giỏi nhưng có sự phân hóa rất mạnh theo vị thế trường (`Tier`), đi kèm biến động hàng năm về tổng chỉ tiêu và tỷ lệ phần trăm chỉ tiêu dành riêng cho phương thức xét điểm thi THPT [3].

**Vấn đề cốt lõi cần giải quyết**

Sau khi Bộ GD&ĐT công bố điểm thi THPT, thí sinh phải đăng ký và sắp xếp thứ tự nguyện vọng trong điều kiện không biết số lượng đối thủ cùng đăng ký vào từng ngành. Thói quen phổ biến của thí sinh là lấy điểm thi năm hiện tại đối chiếu trực tiếp với điểm chuẩn năm trước, dẫn đến hai rủi ro lớn: trượt các nguyện vọng ưu tiên ở nhóm ngành công nghệ mũi nhọn khi tỷ lệ điểm giỏi toàn quốc tăng hoặc nhà trường thu hẹp chỉ tiêu THPT; hoặc đặt nguyện vọng quá an toàn gây lãng phí điểm số khi đề thi khó hơn. Vì vậy, dự án xây dựng hệ thống dữ liệu từ CSDL quan hệ SQL Server đến mô hình Học máy để phân tích các yếu tố chi phối, dự đoán điểm chuẩn năm 2026 và đánh giá mức độ cạnh tranh cho từng chương trình đào tạo thuộc khối ngành Công nghệ – Kỹ thuật tại Hà Nội.

**Giá trị/Tầm quan trọng thực tiễn của dự án**

1. **Làm rõ các yếu tố liên quan đến điểm chuẩn và áp lực cạnh tranh:**

Kết quả phân tích giúp thí sinh, phụ huynh và bộ phận tư vấn tuyển sinh hiểu rõ mối liên hệ giữa điểm chuẩn lịch sử, tỷ lệ thí sinh đạt ngưỡng điểm giỏi (từ 25 và 27 điểm trở lên), áp lực tỷ lệ chọi trên chỉ tiêu THPT thực tế, phân hạng uy tín trường (`Tier`) và độ quan tâm tìm kiếm ngành học (`Google Trends`) đối với mức điểm trúng tuyển.

2. **Hỗ trợ ước tính điểm chuẩn, phân tầng cạnh tranh và sắp xếp nguyện vọng:**

Mô hình cung cấp điểm chuẩn dự báo năm 2026 và phân tầng mức độ cạnh tranh cho từng chương trình đào tạo, giúp thí sinh đối chiếu với điểm thi cá nhân để phân bổ danh sách đăng ký hợp lý vào 3 nhóm (*Nhóm An Toàn*, *Nhóm Vừa Sức*, *Nhóm Thử Thách*). Mức điểm dự đoán đóng vai trò nguồn tham khảo định lượng từ dữ liệu lịch sử và phổ điểm, không thay thế quyết định công bố điểm chuẩn chính thức của Hội đồng tuyển sinh các trường.

3. **Hỗ trợ tra cứu và trực quan hóa dữ liệu tuyển sinh qua Dashboard:**

Ứng dụng Web Streamlit giúp người dùng lọc đa chiều theo trường, nhóm ngành công nghệ, tổ hợp xét tuyển và trực quan hóa sự biến động điểm chuẩn – mức độ cạnh tranh, giảm đáng kể thời gian tổng hợp thủ công từ nhiều đề án tuyển sinh rời rạc.

### 1.2. Các câu hỏi nghiên cứu cốt lõi

- **Câu hỏi 1 (RQ1):** Trong các nhóm yếu tố gồm điểm chuẩn lịch sử, sự dịch chuyển phổ điểm thi THPT toàn quốc (điểm trung bình tổ hợp và tỷ lệ thí sinh đạt từ 25, 27 điểm trở lên), áp lực tỷ lệ chọi trên chỉ tiêu THPT thực tế, xếp hạng trường (`Tier`) và độ quan tâm tìm kiếm (`Google Trends`), những yếu tố nào có mối liên hệ mạnh nhất với điểm chuẩn và mức độ cạnh tranh của khối ngành Công nghệ – Kỹ thuật?
- **Câu hỏi 2 (RQ2):** Có thể dự đoán điểm chuẩn xét tuyển THPT (quy chuẩn thang 30) cho từng chương trình đào tạo thuộc khối Công nghệ – Kỹ thuật tại Hà Nội với sai số trung bình (`MAE`, `RMSE`) bao nhiêu điểm, và các đặc trưng đầu vào giải thích được bao nhiêu phần biến động của điểm chuẩn (`R²`) trên tập kiểm thử độc lập năm 2025?
- **Câu hỏi 3 (RQ3):** Quan hệ giữa các đặc trưng tuyển sinh và điểm chuẩn có mang tính phi tuyến và phân hóa theo phân nhóm kỹ thuật hay không, tức là kiến trúc chia nhóm chuyên gia kết hợp hai nhánh mô hình cây tăng cường `HistGradientBoostingRegressor` (dự đoán trực tiếp mức điểm + dự đoán độ lệch so với năm trước) có dự báo tốt hơn mô hình hồi quy tuyến tính nền (`Linear Regression`) hay không?

### 1.3. Xây dựng vấn đề trong Khoa học Dữ liệu và Phương pháp tiếp cận

**Loại bài toán**

Bài toán Hồi quy (Regression) dự đoán điểm chuẩn kết hợp phân tầng Mức độ cạnh tranh tuyển sinh, tích hợp quy trình quản trị CSDL quan hệ (SQL Server), phân tích khám phá bằng SQL & Python (EDA) và triển khai ứng dụng Dashboard tư vấn nguyện vọng.

**Biến mục tiêu (y) và Đầu ra đánh giá mức độ cạnh tranh**

- **Biến mục tiêu hồi quy (`diem_chuan`):** Điểm chuẩn trúng tuyển theo phương thức xét điểm thi tốt nghiệp THPT của từng chương trình đào tạo khối Công nghệ – Kỹ thuật tại Hà Nội, được quy đổi đồng nhất về thang điểm 30,0. Trong nhánh mô hình thứ hai (`Delta`), biến mục tiêu trung gian là biên độ thay đổi điểm chuẩn so với năm liền trước ($\Delta y_t = y_t - y_{t-1}$), sau đó cộng vào điểm chuẩn năm trước để đưa về cùng thang điểm 30,0.
- **Nhãn mức độ cạnh tranh đầu ra (`muc_do_canh_tranh`):- **Chỉ số và Nhãn mức độ cạnh tranh đầu ra (`chi_so_canh_tranh`, `muc_do_canh_tranh`):** Thay vì chỉ cắt ngưỡng đơn biến theo điểm chuẩn, hệ thống tổng hợp **Chỉ số Cạnh tranh Tuyển sinh (`CI`, thang 0 – 100)** từ 4 yếu tố: *Ngưỡng điểm chuẩn dự báo 2026 (50%)* + *Áp lực tỷ lệ chọi thực tế trên chỉ tiêu THPT `ti_le_choi_tong` (25%)* + *Vị thế trường `tier_score` (15%)* + *Đà tăng điểm chuẩn $\Delta$ so với năm trước (10%)*. Từ chỉ số `CI`, các ngành được phân tầng thành 4 cấp độ: *Rất cao - Khốc liệt* ($\text{CI} \ge 72$), *Cao* ($60 \le \text{CI} < 72$), *Trung bình* ($48 \le \text{CI} < 60$) và *Thấp - Dễ trúng tuyển* ($\text{CI} < 48$).
**Các biến đặc trưng dự kiến (X)**

- **Đặc điểm lịch sử điểm chuẩn:** Điểm chuẩn năm liền trước của chính ngành đó (`diem_nam_ngoai`), điểm chuẩn trung bình năm trước của toàn trường (`school_avg_score`) và điểm chuẩn trung bình năm trước của phân nhóm ngành công nghệ (`major_avg_score`).
- **Đặc điểm phổ điểm thi THPT toàn quốc:** Điểm thi trung bình toàn quốc của khối xét tuyển sau khi cộng 0,5 điểm ưu tiên quy chuẩn (`Diem_Trung_Binh`), tỷ lệ thí sinh đạt ngưỡng điểm giỏi từ 25 và 27 điểm trở lên (`ti_le_tren_25`, `ti_le_tren_27`), các biến trễ phổ điểm của năm trước (`diem_tb_nam_ngoai`, `ti_le_25_nam_ngoai`, `ti_le_27_nam_ngoai`), tổng quy mô thí sinh (`Tong_So_Thi_Sinh`) và tỷ số giữa điểm chuẩn năm trước với mặt bằng đề thi (`ratio_diem_pho_diem`).
- **Đặc điểm chỉ tiêu và áp lực tỷ lệ chọi:** Số suất chỉ tiêu thực tế dành cho xét điểm thi THPT (`chi_tieu_thuc_te`) và áp lực tỷ lệ chọi thực tế tính theo hàm logarit giữa số lượng thí sinh đạt điểm cao của tổ hợp trên chỉ tiêu ngành (`ti_le_choi_tong`).
- **Đặc điểm trường, phân nhóm ngành, xu hướng và quy chế:** Mã trường đại học (`ma_truong_cat`), mã phân nhóm ngành chuẩn (`ma_nganh_chuan_cat` thuộc 5 phân nhóm chuyên sâu của hai mảng `IT` và `ENG`: `AI_DATA`, `SE_CYBER`, `CHIP_IOT_HW`, `AUTO_ELEC`, `MECH_AUTO_TECH`), điểm phân hạng trường (`tier_score`: Tier 1 = 3, Tier 2 = 2, Tier 3 = 1), chỉ số độ hot tìm kiếm trên Google (`trend_score`) [4] và biến cờ đánh dấu quy chế thi mới (`thi_format_moi`).

Toàn bộ biến đầu vào đều là thông tin xác định được *trước khi các trường công bố điểm chuẩn của năm dự báo*, đảm bảo nguyên tắc chống rò rỉ dữ liệu (Data Leakage) [5]. Từ 18 cột đặc trưng chế biến ở Giai đoạn 4, mô hình ở Giai đoạn 5 đưa vào huấn luyện 16 đặc trưng cốt lõi (lược bỏ `Tong_So_Thi_Sinh` dạng số đếm tuyệt đối và `thi_format_moi` để tránh nhiễu quy mô mẫu khi chuyển giao quy chế thi 2025–2026).

**Tiêu chí đánh giá**

- **MAE (Mean Absolute Error):** Đo sai số tuyệt đối trung bình giữa điểm chuẩn thực tế và điểm chuẩn dự đoán trên thang điểm 30 (đơn vị: điểm). Đây là chỉ số đánh giá chính vì phản ánh trực tiếp độ lệch trung bình khi tư vấn cho thí sinh; giá trị càng thấp càng tốt.
- **RMSE (Root Mean Squared Error):** Đo căn bậc hai trung bình bình phương sai số (đơn vị: điểm), phạt nặng các trường hợp ngành học có điểm chuẩn biến động đột biến giữa hai năm; giá trị càng thấp càng tốt.
- **R² (R-squared):** Đánh giá tỷ lệ biến thiên của điểm chuẩn giữa các trường và chương trình đào tạo được mô hình giải thích; giá trị càng cao càng tốt.

**Phương pháp đánh giá mô hình**

Do mặt bằng điểm chuẩn từng năm chịu tác động chung từ độ khó đề thi THPT của năm đó, việc chia ngẫu nhiên theo dòng sẽ làm lộ thông tin mặt bằng điểm năm tương lai vào quá trình huấn luyện. Vì vậy, nhóm chia dữ liệu nghiêm ngặt theo thời gian (**Temporal Split**): sử dụng giai đoạn **2023–2024** làm **Tập Học** ($n = 262$ chương trình có lịch sử điểm trễ hợp lệ) và toàn bộ năm **2025** làm **Tập Kiểm Thử độc lập** ($n = 151$ chương trình), cố định `random_state = 42` để đảm bảo tính lặp lại trước khi dự báo cho **216 chương trình** năm **2026**.

**Phạm vi và hạn chế của nghiên cứu**

Nghiên cứu tập trung vào các chương trình đào tạo thuộc khối ngành Công nghệ – Kỹ thuật tại khu vực Hà Nội xét tuyển bằng điểm thi tốt nghiệp THPT trên thang điểm 30. Kết quả mô hình không mặc nhiên áp dụng cho các khối ngành Kinh tế, Luật, Y Dược, Xã hội hoặc các phương thức xét tuyển riêng (Đánh giá tư duy, Đánh giá năng lực, xét tuyển chứng chỉ quốc tế).

Bên cạnh đó, giai đoạn 2025–2026 có sự thay đổi quy chế thi sang 4 môn và xuất hiện 65 chương trình đào tạo mới mở hoặc đổi tên năm 2025 chưa có chuỗi lịch sử dài hạn. Ngoài ra, dữ liệu công khai không bao gồm số lượng hồ sơ đăng ký thực tế ở từng thứ tự nguyện vọng trên hệ thống lọc ảo của Bộ GD&ĐT cũng như điểm ưu tiên riêng của từng thí sinh (được cộng quy chuẩn 0,5 điểm khi tính phổ điểm khối), nên mô hình chỉ lượng hóa quy luật từ dữ liệu lịch sử quan sát được mà không khẳng định quan hệ nhân quả tuyệt đối.
